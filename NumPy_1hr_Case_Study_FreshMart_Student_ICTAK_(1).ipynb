### Task 6 · Fill the gaps
Create a copy `clean` of `sales` in which every missing value is replaced by the **average of its own department** (ignoring missing values). Print the number of missing values left and the department averages used (1 decimal).

**Expected output**
```
0
[60.7 34.  19.3 29.7 25.3]
```

````python
# Create a copy of the sales array
clean = sales.copy()
dept_averages = np.nanmean(clean, axis=0)
nan_mask = np.isnan(clean)
clean[nan_mask] = np.take(dept_averages, np.where(nan_mask)[1])
print(np.isnan(clean).sum())
print(np.round(dept_averages, 1))
````

### Task 9 · Best day of the week
Add the departments together to get the sales of the whole store for each of the 28 days. Then find the **average store sales for each weekday** (Mon to Sun) across the 4 weeks. Print the 7 averages (1 decimal) and the name of the best weekday.

### Task 12 · Busy days
A day is **busy** if the store sales for that day are more than **1.2 times** the average daily store sales. Print the number of busy days and their day numbers (1 to 28).

**Expected output**

### Task 14 · Department ranking
Rank the departments from the highest to the lowest 4-week total. Print their names in that order, and then the **share of total store sales** of each department in that order (percentage, 1 decimal).

**Expected output**
```
['Grocery' 'Dairy' 'Fruits' 'Snacks' 'Bakery']
[35.7 20.  17.5 14.9 11.8]
```

````python

dept_totals = clean.sum(axis=0)


sorted_indices = np.argsort(dept_totals)[::-1]
ranked_departments = departments[sorted_indices]


total_store_sales = clean.sum()
ranked_shares = (dept_totals[sorted_indices] / total_store_sales) * 100


print(ranked_departments)
print(np.round(ranked_shares, 1))
````

### Task 16 · 7-day moving average
A **moving average** smooths daily numbers. For `store_daily`, calculate the average of every 7 consecutive days (days 1–7, days 2–8, and so on). Print how many averages there are and the first 3 (1 decimal).

**Expected output**
```
22
[163.2 164.4 165.4]
```

````python
window_size = 7
moving_averages = np.convolve(store_daily, np.ones(window_size) / window_size, mode='valid')


print(len(moving_averages))
print(np.round(moving_averages[:3], 1))
````
